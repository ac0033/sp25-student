## Evaluating Classifiers

In our models, we evaluate the accuracy of the training set, which may provide a misleading accuracy measure. In Project B1, we calculated various metrics to consider other ways of evaluating a classifier, in addition to overall accuracy. Below is a reference to those concepts.

Presumably, our classifier will be used for **filtering**, or preventing messages labeled `spam` from reaching someone's inbox. There are two kinds of errors we can make:
- **False positive (FP)**: A ham email gets flagged as spam and filtered out of the inbox.
- **False negative (FN)**: A spam email gets mislabeled as ham and ends up in the inbox.

To be clear, we label spam emails as 1 and ham emails as 0. These definitions depend both on the true labels and the predicted labels. False positives and false negatives may be of differing importance, leading us to consider more ways of evaluating a classifier in addition to overall accuracy:

**Precision**: Measures the proportion of emails flagged as spam that are actually spam. Mathematically, $\frac{\text{TP}}{\text{TP} + \text{FP}}$.

**Recall**: Measures the proportion  of spam emails that were correctly flagged as spam. Mathematically, $\frac{\text{TP}}{\text{TP} + \text{FN}}$.

**False positive rate**: Measures the proportion  of ham emails that were incorrectly flagged as spam. Mathematically, $\frac{\text{FP}}{\text{FP} + \text{TN}}$.

```python
def extract_html_features(email_text):
    features = {}
    
    html_tags = ['<html>', '<head>', '<body>', '<table>', '<center>', '<font>', '<a>', '<img>', '<br>']
    for tag in html_tags:
        tag_name = tag.replace("<", "").replace(">", "")
        features[f'has_html_{tag_name}'] = int(tag in email_text.lower())
    
    features['html_tag_count'] = email_text.lower().count('<')
    features['has_any_html'] = int(features['html_tag_count'] > 0)
    
    return features
```

```python
def extract_length_features(email_text, subject_text):
    features = {}
    
    features['email_char_length'] = len(email_text)
    features['subject_char_length'] = len(subject_text)
    features['email_word_count'] = len(email_text.split())
    features['subject_word_count'] = len(subject_text.split())
    
    features['email_to_subject_char_ratio'] = features['email_char_length'] / max(features['subject_char_length'], 1)
    features['email_to_subject_word_ratio'] = features['email_word_count'] / max(features['subject_word_count'], 1)
    
    return features
```

```python
def extract_special_char_features(email_text, subject_text):
    features = {}
    
    features['subject_caps_ratio'] = sum(1 for c in subject_text if c.isupper()) / max(len(subject_text), 1)
    features['email_caps_ratio'] = sum(1 for c in email_text if c.isupper()) / max(len(email_text), 1)
    
    special_chars = ['!', '?', '$', '%', '#', '@', '&', '*']
    for char in special_chars:
        features[f'subject_{char}_count'] = subject_text.count(char)
        features[f'email_{char}_count'] = email_text.count(char)
    
    features['subject_digit_count'] = sum(1 for c in subject_text if c.isdigit())
    features['email_digit_count'] = sum(1 for c in email_text if c.isdigit())
    features['subject_digit_ratio'] = features['subject_digit_count'] / max(len(subject_text), 1)
    features['email_digit_ratio'] = features['email_digit_count'] / max(len(email_text), 1)
    
    return features
```

```python
def extract_spam_keywords(email_text, subject_text):
    spam_keywords = [
        'free', 'money', 'offer', 'urgent', 'win', 'guarantee', 'click here',
        'unsubscribe', 'sex', 'viagra', 'loan', 'credit', 'debt', 'cash',
        'prize', 'winner', 'congratulations', 'limited time', 'act now',
        'no obligation', 'risk free', '100% free', 'save money', 'earn money',
        'make money', 'work from home', 'get rich', 'investment', 'profit'
    ]
    
    features = {}
    combined_text = (subject_text + ' ' + email_text).lower()
    
    for keyword in spam_keywords:
        feature_name = f'has_spam_keyword_{keyword.replace(" ", "_").replace("%", "percent")}'
        features[feature_name] = int(keyword in combined_text)
    
    features['total_spam_keywords'] = sum(features[f'has_spam_keyword_{keyword.replace(" ", "_").replace("%", "percent")}'] for keyword in spam_keywords)
    features['spam_keyword_density'] = features['total_spam_keywords'] / max(len(combined_text.split()), 1)
    
    return features
```

```python
def extract_structure_features(email_text):
    features = {}
    
    features['has_mime_boundary'] = int('--===' in email_text or 'boundary=' in email_text)
    
    features['has_repeated_chars'] = int('#' * 10 in email_text or '=' * 10 in email_text)
    
    features['has_http_links'] = int('http://' in email_text.lower() or 'https://' in email_text.lower())
    
    features['has_from_header'] = int(email_text.lower().startswith('from:'))
    features['has_to_header'] = int('to:' in email_text.lower())
    
    features['newline_count'] = email_text.count('\n')
    features['newline_density'] = features['newline_count'] / max(len(email_text), 1)
    
    return features
```

```python
def extract_all_features(email_text, subject_text):
    features = {}
    
    features.update(extract_html_features(email_text))
    features.update(extract_length_features(email_text, subject_text))
    features.update(extract_special_char_features(email_text, subject_text))
    features.update(extract_spam_keywords(email_text, subject_text))
    features.update(extract_structure_features(email_text))
    
    return features
```

```python
from sklearn.preprocessing import StandardScaler

def create_feature_dataframe(data):
    feature_list = []
    
    for _, row in data.iterrows():
        features = extract_all_features(row['email'], row['subject'])
        feature_list.append(features)
    
    feature_df = pd.DataFrame(feature_list)
    
    feature_df = feature_df.fillna(0)
    return feature_df

test = test.fillna('')
X_train_features_df = create_feature_dataframe(train)
X_val_features_df = create_feature_dataframe(val)
X_test_features_df = create_feature_dataframe(test)
```

```python
def l1_feature_selection(X_train_features_df, Y_train, C_values=None):
    if C_values is None:
        C_values = [0.001, 0.01, 0.1, 1.0, 10.0]
    
    results = []

    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_train_features_df)
    
    for C in C_values:
        model = LogisticRegression(
            C=C, 
            penalty='l1', 
            solver='liblinear', 
            random_state=42,
            max_iter=1000
        )
        model.fit(X_scaled, Y_train)
        
        selected_features = X_train_features_df.columns[model.coef_[0] != 0].tolist()
        num_features = len(selected_features)
        
        results.append({
            'C': C,
            'num_features': num_features,
            'selected_features': selected_features,
            'coefficients': model.coef_[0]
        })
    
    return results


l1_results = l1_feature_selection(X_train_features_df, Y_train)
```

```python
def select_optimal_features(l1_results, X_train_features_df, Y_train):    
    target_features = 20  
    best_result = None
    min_diff = float('inf')
    
    for result in l1_results:
        diff = abs(result['num_features'] - target_features)
        if diff < min_diff:
            min_diff = diff
            best_result = result
    
    selected_features = best_result['selected_features']
    X_selected = X_train_features_df[selected_features]
    
    return X_selected, selected_features

X_selected, selected_features = select_optimal_features(l1_results, X_train_features_df, Y_train)
X_val_selected = X_val_features_df[selected_features]
X_test_selected = X_test_features_df[selected_features]
```

```python
from scipy.stats import chi2_contingency, pearsonr

def statistical_feature_analysis(X_train_features_df, Y_train):
    results = []
    
    for feature in X_train_features_df.columns:
        if X_train_features_df[feature].nunique() == 2:
            contingency_table = pd.crosstab(X_train_features_df[feature], Y_train)
            chi2, p_value, _, _ = chi2_contingency(contingency_table)
            results.append({
                'feature': feature,
                'test': 'chi2',
                'statistic': chi2,
                'p_value': p_value,
                'significant': p_value < 0.05
            })
        else:
            corr, p_value = pearsonr(X_train_features_df[feature], Y_train)
            results.append({
                'feature': feature,
                'test': 'pearson',
                'statistic': corr,
                'p_value': p_value,
                'significant': p_value < 0.05
            })
    
    results_df = pd.DataFrame(results)
    results_df = results_df.sort_values('statistic', key=abs, ascending=False)
    
    return results_df

stat_results = statistical_feature_analysis(X_selected, Y_train)
selected_features_update = stat_results['feature'][stat_results['significant']]
X_selected_update = X_selected[selected_features_update]
X_val_selected_update = X_val_selected[selected_features_update]
X_test_selected_update = X_test_selected[selected_features_update]
```

```python
def analyze_multicollinearity(X_train_features_df, threshold=0.8):
    corr_matrix = X_train_features_df.corr()
    
    high_corr_pairs = []
    for i in range(len(corr_matrix.columns)):
        for j in range(i+1, len(corr_matrix.columns)):
            corr_value = abs(corr_matrix.iloc[i, j])
            if corr_value > threshold:
                high_corr_pairs.append({
                    'feature1': corr_matrix.columns[i],
                    'feature2': corr_matrix.columns[j],
                    'correlation': corr_value
                })
    
    high_corr_df = pd.DataFrame(high_corr_pairs)
    high_corr_df = high_corr_df.sort_values('correlation', ascending=False)
    
    plt.figure(figsize=(15, 12))
    mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
    sns.heatmap(corr_matrix, mask=mask, annot=False, cmap='coolwarm', center=0)
    plt.title('Feature Correlation Matrix')
    plt.tight_layout()
    plt.show()

analyze_multicollinearity(X_selected_update, threshold=0.7)
```

# Question 5: ROC Curve

In most cases, we won't be able to get 0 false positives and 0 false negatives, so we have to compromise. For example, in the case of cancer screenings, false negatives are comparatively worse than false positives — a false negative means that a patient might not discover that they have cancer until it's too late. In contrast, a patient can receive another screening for a false positive.

Recall that logistic regression calculates the probability that an example belongs to a particular class. To classify an example, we say that an email is spam if our classifier gives it $\ge 0.5$ probability of being spam. However, **we can adjust that cutoff threshold**. We can say that an email is spam only if our classifier gives it $\ge 0.7$ probability of being spam, for example. This is how we can trade off false positives and false negatives.

The Receiver Operating Characteristic (ROC) curve shows this trade-off for each possible cutoff probability. In the cell below, plot an ROC curve for your final classifier (the one you use to make predictions for Gradescope) on the training data. [Lecture 23](https://ds100.org/fa24/lecture/lec23/) may be helpful.

**Hint**: You'll want to use the `.predict_proba` method [(documentation)](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LogisticRegression.html#sklearn.linear_model.LogisticRegression.predict_proba) for your classifier instead of `.predict` to get probabilities instead of binary predictions.

```python
fpr, tpr, thresholds = roc_curve(Y_train, my_model.predict_proba(X_train_scaled)[:, 1])

roc_auc = auc(fpr, tpr)

plt.figure(figsize=(10, 8))
plt.plot(fpr, tpr, color='darkorange', lw=3, 
         label=f'ROC Curve (AUC = {roc_auc:.3f})')

plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', 
         label='Random Classifier (AUC = 0.500)')

plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])

plt.xlabel('False Positive Rate (1 - Specificity)', fontsize=12, fontweight='bold')
plt.ylabel('True Positive Rate (Sensitivity)', fontsize=12, fontweight='bold')
plt.title('Receiver Operating Characteristic (ROC) Curve\nSpam Email Classification', 
          fontsize=14, fontweight='bold', pad=20)

plt.legend(loc="lower right", fontsize=11, framealpha=0.9)

plt.grid(True, alpha=0.3, linestyle='-', linewidth=0.5)

distances = np.sqrt((fpr - 0)**2 + (tpr - 1)**2)
best_idx = np.argmin(distances)
best_fpr, best_tpr = fpr[best_idx], tpr[best_idx]
best_threshold = thresholds[best_idx]

plt.plot(best_fpr, best_tpr, 'ro', markersize=8, 
         label=f'Best Threshold = {best_threshold:.3f}')
plt.annotate(f'({best_fpr:.3f}, {best_tpr:.3f})', 
             xy=(best_fpr, best_tpr), 
             xytext=(best_fpr + 0.1, best_tpr + 0.1),
             arrowprops=dict(arrowstyle='->', color='red', alpha=0.7),
             fontsize=10, color='red')

plt.tight_layout()
plt.show()
```
